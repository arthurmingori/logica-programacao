# Estruturas de Repetição

# For

```python
# For Range
for contador in range(10):
  print(contador)
```

```python
# For Range + Break
for contador in range(1000):
  print(contador)
  if contador == 2:
    break
```

```python
lista = ['teste1', 'teste2']
```

```python
# For Simples
for val in lista:
  print(val)
```

```python
# For com Enumerate
for i, val in enumerate(lista):
  print(i, val)
```

# While

```python
val = 1
while val < 6:
  val *=2
  print(val)
```

```python
# While com else
val = 10
while val < 6:
  val *=2
  print(val)

else:
  print('Valor de val é:', val)
```

# Exercícios

Calcular a média entre duas notas

```python
nota1 = input('Digite a primeira nota ')
nota2 = input('Digite a segunda nota ')
nota1_convert = float(nota1)
nota2_convert = float(nota2)
media = (nota1_convert + nota2_convert) /2
print('A média das notas é,', media)
```

Verificação de senha

```python
senha_permitida = 12345
senha_digitada = input('Digite a senha')
senha_permitida = int(senha_permitida)
senha_digitada = int(senha_digitada)

if senha_permitida == senha_digitada:
  print('Acesso permitido')

else:
  print('Acesso negado')
```

Lista de notas

```python
# Lista de nota dos alunos

notas = [77, 93, 65, 48, 55, 82, 75]

# Calcule a media das notas

soma = 0

for nota in notas:
  soma += nota
  media = soma / len(notas) # len é utilizado para retomar o número de notas nesse caso são 7

# Classifica os alunos com base na media

alunos_aprovados = []
alunos_reprovados = []

for nota in notas:
  if nota >= media:
    alunos_aprovados.append(nota)

  else:
    alunos_reprovados.append(nota)

# Imprime os resultados

print('Notas dos alunos', notas)
print('Media das notas', media)
print('Alunos aprovados', alunos_aprovados)
print('Alunos reprovados', alunos_reprovados)
```
